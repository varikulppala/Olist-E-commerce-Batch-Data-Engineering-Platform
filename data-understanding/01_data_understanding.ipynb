```python
import pandas as pd
import os
from pathlib import Path
```

```python
pd.set_option("display.max_columns", None)
pd.set_option("display.width", 200)
```

```python
DATA_PATH = Path("C:\\Users\\varik\\OneDrive\\Desktop\\aws-batch-data-platform\\data\\raw")

print("Dataset path:", DATA_PATH.resolve())
print("\nFiles found:")

for file in DATA_PATH.glob("*.csv"):
    print(file.name)
```

```python
customers = pd.read_csv(DATA_PATH / "olist_customers_dataset.csv")
geolocation = pd.read_csv(DATA_PATH / "olist_geolocation_dataset.csv")
order_items = pd.read_csv(DATA_PATH / "olist_order_items_dataset.csv")
payments = pd.read_csv(DATA_PATH / "olist_order_payments_dataset.csv")
reviews = pd.read_csv(DATA_PATH / "olist_order_reviews_dataset.csv")
orders = pd.read_csv(DATA_PATH / "olist_orders_dataset.csv")
products = pd.read_csv(DATA_PATH / "olist_products_dataset.csv")
sellers = pd.read_csv(DATA_PATH / "olist_sellers_dataset.csv")
category_translation = pd.read_csv(
    DATA_PATH / "product_category_name_translation.csv"
)
```

```python
datasets = {
    "customers": customers,
    "geolocation": geolocation,
    "order_items": order_items,
    "payments": payments,
    "reviews": reviews,
    "orders": orders,
    "products": products,
    "sellers": sellers,
    "category_translation": category_translation
}
```

```python
summary = []

for name, df in datasets.items():
    summary.append({
        "dataset": name,
        "rows": df.shape[0],
        "columns": df.shape[1],
        "memory_mb": round(df.memory_usage(deep=True).sum() / (1024 * 1024), 2)
    })

dataset_summary = pd.DataFrame(summary)

dataset_summary
```

```python
dataset_summary.to_csv(
    "../docs/dataset_inventory.csv",
    index=False
)
```

```python
for name, df in datasets.items():
    print("=" * 100)
    print(f"DATASET: {name.upper()}")
    print("=" * 100)

    print("\nShape:")
    print(df.shape)

    print("\nColumns:")
    print(df.columns.tolist())

    print("\nData Types:")
    print(df.dtypes)

    print("\nFirst 3 Rows:")
    display(df.head(3))
```

```python
null_summary = []

for name, df in datasets.items():
    for column in df.columns:
        null_count = df[column].isnull().sum()

        null_summary.append({
            "dataset": name,
            "column": column,
            "null_count": null_count,
            "null_percentage": round(
                (null_count / len(df)) * 100, 2
            )
        })

null_report = pd.DataFrame(null_summary)

null_report = null_report[
    null_report["null_count"] > 0
].sort_values(
    ["dataset", "null_percentage"],
    ascending=[True, False]
)

null_report
```

```python
null_report.to_csv(
    "../docs/null_value_report.csv",
    index=False
)
```

```python
duplicate_summary = []

for name, df in datasets.items():
    duplicate_count = df.duplicated().sum()

    duplicate_summary.append({
        "dataset": name,
        "total_rows": len(df),
        "duplicate_rows": duplicate_count,
        "duplicate_percentage": round(
            duplicate_count / len(df) * 100, 2
        )
    })

duplicate_report = pd.DataFrame(duplicate_summary)

duplicate_report
```

```python
duplicate_report.to_csv(
    "../docs/duplicate_report.csv",
    index=False
)
```
